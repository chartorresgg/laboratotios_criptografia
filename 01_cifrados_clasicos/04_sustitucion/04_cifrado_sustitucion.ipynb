````python
# ============================================
# CONFIGURACIÓN - Modifica aquí
# ============================================

MENSAJE = "HELLO WORLD"           # Mensaje a cifrar

# Clave: permutación del alfabeto
# Original: ABCDEFGHIJKLMNOPQRSTUVWXYZ
CLAVE = "QWERTYUIOPASDFGHJKLZXCVBNM"

ALFABETO_NORMAL = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'

print(f"📝 Mensaje: {MENSAJE}")
print(f"🔑 Clave:   {CLAVE}")
print(f"\nAlfabeto normal: {ALFABETO_NORMAL}")
print(f"Alfabeto clave:  {CLAVE}")
````

### 1. Función: Generar Clave Aleatoria

````python
import random
from collections import Counter

def generar_clave_aleatoria():
    """
    Genera una clave aleatoria como permutación del alfabeto.
    
    Una permutación significa que cada letra aparece exactamente una vez,
    pero en un orden aleatorio diferente.
    
    Proceso:
    1. Toma todas las letras del alfabeto
    2. Las mezcla aleatoriamente (shuffle)
    3. Las une en una cadena
    
    Ejemplo de salidas posibles:
        "QWERTYUIOPASDFGHJKLZXCVBNM"
        "ZXCVBNMASDFGHJKLQWERTYUIOP"
        "ZYXWVUTSRQPONMLKJIHGFEDCBA"
    """
    letras = list(ALFABETO_NORMAL)
    random.shuffle(letras)
    return ''.join(letras)

# Generar algunas claves de ejemplo
print("Ejemplos de claves aleatorias:")
print("-"*50)
for i in range(5):
    clave_ejemplo = generar_clave_aleatoria()
    print(f"{i+1}. {clave_ejemplo}")

print(f"\n💡 Cada clave es única y tiene 26! ≈ 4×10²⁶ posibilidades")
````

### 2. Función: Validar Clave

````python
def validar_clave(clave):
    """
    Valida que la clave sea una permutación válida del alfabeto.
    
    Requisitos para una clave válida:
    1. Debe tener exactamente 26 letras
    2. Cada letra del alfabeto debe aparecer exactamente una vez
    3. No puede haber letras duplicadas
    4. No puede faltar ninguna letra
    
    Proceso de validación:
    1. Limpiar la clave (solo letras, mayúsculas)
    2. Verificar longitud = 26
    3. Contar cada letra (debe ser exactamente 1)
    4. Verificar que no falte ninguna letra del alfabeto
    
    Retorna:
        Tupla (es_valida, mensaje)
    """
    # Limpiar clave
    clave_limpia = ''.join(c.upper() for c in clave if c.isalpha())
    
    # Verificar longitud
    if len(clave_limpia) != 26:
        return False, f"Longitud incorrecta: {len(clave_limpia)} (debe ser 26)"
    
    # Contar frecuencias
    contador = Counter(clave_limpia)
    
    # Buscar letras duplicadas
    letras_duplicadas = [letra for letra, count in contador.items() if count > 1]
    if letras_duplicadas:
        return False, f"Letras duplicadas: {', '.join(letras_duplicadas)}"
    
    # Buscar letras faltantes
    letras_faltantes = [letra for letra in ALFABETO_NORMAL if letra not in clave_limpia]
    if letras_faltantes:
        return False, f"Letras faltantes: {', '.join(letras_faltantes)}"
    
    return True, "✓ Clave válida"

# Probar validación
print("Pruebas de validación:")
print("="*70)

pruebas = [
    (CLAVE, "Clave configurada"),
    ("ABCDEFGHIJKLMNOPQRSTUVWXYZ", "Alfabeto normal"),
    ("AAABCDEFGHIJKLMNOPQRSTUVWXYZ", "Con duplicados"),
    ("ABCDEFGHIJKLMNOPQRSTUVWXY", "Falta la Z"),
    ("HELLO", "Muy corta")
]

for clave_test, descripcion in pruebas:
    valida, mensaje = validar_clave(clave_test)
    estado = "✓" if valida else "✗"
    print(f"{estado} {descripcion:25} → {mensaje}")
````

### 3. Función: Crear Tabla de Sustitución

````python
def crear_tabla_sustitucion(clave):
    """
    Crea las tablas de mapeo para cifrado y descifrado.
    
    El cifrado de sustitución requiere dos tablas:
    1. Tabla de cifrado: alfabeto normal → clave
    2. Tabla de descifrado: clave → alfabeto normal (inversa)
    
    Proceso:
    1. Limpiar la clave
    2. Crear diccionario mapeando cada letra normal a su sustituto
    3. Crear diccionario inverso para descifrado
    
    Ejemplo:
        Si CLAVE = "QWERTYUIOPASDFGHJKLZXCVBNM"
        
        Tabla de cifrado:
        A → Q, B → W, C → E, ...
        
        Tabla de descifrado:
        Q → A, W → B, E → C, ...
    
    Retorna:
        Tupla (tabla_cifrado, tabla_descifrado)
    """
    clave_limpia = ''.join(c.upper() for c in clave if c.isalpha())
    
    # Tabla para cifrar: normal → clave
    tabla_cifrado = {}
    for i, letra_normal in enumerate(ALFABETO_NORMAL):
        tabla_cifrado[letra_normal] = clave_limpia[i]
    
    # Tabla para descifrar: clave → normal (inversa)
    tabla_descifrado = {}
    for i, letra_normal in enumerate(ALFABETO_NORMAL):
        tabla_descifrado[clave_limpia[i]] = letra_normal
    
    return tabla_cifrado, tabla_descifrado

# Crear tablas para nuestra clave
tabla_cifrado, tabla_descifrado = crear_tabla_sustitucion(CLAVE)

print("TABLA DE SUSTITUCIÓN")
print("="*70)
print("\nTabla de Cifrado (Original → Cifrado):")
print("-"*70)
for i in range(0, 26, 13):  # Mostrar en 2 filas
    for j in range(i, min(i+13, 26)):
        letra_normal = ALFABETO_NORMAL[j]
        letra_cifrada = tabla_cifrado[letra_normal]
        print(f"{letra_normal}→{letra_cifrada}", end="  ")
    print()

print("\nVisualización completa:")
print(f"Normal:  {ALFABETO_NORMAL}")
print(f"         {'↓ '*26}")
print(f"Cifrado: {CLAVE}")
````

### 4. Función: Cifrar

````python
def limpiar_texto(texto):
    """Limpia el texto dejando solo letras mayúsculas."""
    return ''.join(c.upper() for c in texto if c.isalpha())

def cifrar_sustitucion(texto, clave):
    """
    Cifra un texto usando el Cifrado de Sustitución.
    
    Proceso completo paso a paso:
    1. Validar que la clave sea una permutación válida del alfabeto
    2. Limpiar el texto (solo letras mayúsculas)
    3. Crear tabla de sustitución
    4. Para cada letra del mensaje:
       - Buscar en la tabla su letra correspondiente
       - Reemplazar
    5. Construir texto cifrado
    
    Ejemplo:
        Texto: "HELLO"
        Clave: "QWERTYUIOPASDFGHJKLZXCVBNM"
        
        H → I
        E → Y
        L → S
        L → S
        O → B
        
        Resultado: "IYSSB"
    
    Parámetros:
        texto: mensaje a cifrar
        clave: permutación del alfabeto
    
    Retorna:
        Tupla (texto_cifrado, tabla_cifrado, detalles)
    """
    # Validar clave
    es_valida, mensaje = validar_clave(clave)
    if not es_valida:
        raise ValueError(f"Clave inválida: {mensaje}")
    
    # Limpiar texto
    texto_limpio = limpiar_texto(texto)
    if not texto_limpio:
        raise ValueError("El texto no puede estar vacío")
    
    # Crear tabla
    tabla_cifrado, _ = crear_tabla_sustitucion(clave)
    
    # Cifrar letra por letra
    texto_cifrado = ""
    detalles = []
    
    for i, letra in enumerate(texto_limpio):
        letra_cifrada = tabla_cifrado[letra]
        texto_cifrado += letra_cifrada
        
        detalles.append({
            'pos': i,
            'original': letra,
            'val_orig': ord(letra) - ord('A'),
            'cifrado': letra_cifrada,
            'val_cifr': ord(letra_cifrada) - ord('A')
        })
    
    return texto_cifrado, tabla_cifrado, detalles

# CIFRAR EL MENSAJE
texto_cifrado, tabla_cif, detalles_cif = cifrar_sustitucion(MENSAJE, CLAVE)

print("="*80)
print("PROCESO DE CIFRADO - SUSTITUCIÓN")
print("="*80)
print(f"\nMensaje original: {MENSAJE}")
print(f"Mensaje limpio:   {limpiar_texto(MENSAJE)}")
print(f"Clave:            {CLAVE}")

print("\nTabla de sustitución:")
print(f"Normal:  {ALFABETO_NORMAL}")
print(f"         {'↓ '*26}")
print(f"Cifrado: {CLAVE}")

print("\n" + "-"*80)
print(f"{'Pos':<5} {'Original':<10} {'Val':<5} {'→':<3} {'Cifrado':<10} {'Val':<5}")
print("-"*80)

for d in detalles_cif:
    print(f"{d['pos']:<5} {d['original']:<10} {d['val_orig']:<5} {'→':<3} "
          f"{d['cifrado']:<10} {d['val_cifr']:<5}")

print("\n" + "="*80)
print(f"🔒 MENSAJE CIFRADO: {texto_cifrado}")
print("="*80)
````

### 5. Función: Descifrar

````python
def descifrar_sustitucion(texto_cifrado, clave):
    """
    Descifra un texto usando el Cifrado de Sustitución.
    
    Proceso completo paso a paso:
    1. Validar que la clave sea válida
    2. Limpiar el texto cifrado
    3. Crear tabla de sustitución INVERSA
    4. Para cada letra cifrada:
       - Buscar en la tabla inversa su letra original
       - Reemplazar
    5. Construir texto descifrado
    
    La tabla inversa es clave:
        Si para cifrar: A → Q
        Para descifrar: Q → A
    
    Parámetros:
        texto_cifrado: mensaje cifrado
        clave: permutación usada en el cifrado
    
    Retorna:
        Tupla (texto_descifrado, tabla_descifrado, detalles)
    """
    # Validar clave
    es_valida, mensaje = validar_clave(clave)
    if not es_valida:
        raise ValueError(f"Clave inválida: {mensaje}")
    
    # Limpiar texto cifrado
    texto_limpio = limpiar_texto(texto_cifrado)
    if not texto_limpio:
        raise ValueError("El texto cifrado no puede estar vacío")
    
    # Crear tabla inversa
    _, tabla_descifrado = crear_tabla_sustitucion(clave)
    
    # Descifrar letra por letra
    texto_descifrado = ""
    detalles = []
    
    for i, letra in enumerate(texto_limpio):
        letra_original = tabla_descifrado[letra]
        texto_descifrado += letra_original
        
        detalles.append({
            'pos': i,
            'cifrado': letra,
            'val_cifr': ord(letra) - ord('A'),
            'original': letra_original,
            'val_orig': ord(letra_original) - ord('A')
        })
    
    return texto_descifrado, tabla_descifrado, detalles

# DESCIFRAR EL MENSAJE
texto_descifrado, tabla_desc, detalles_desc = descifrar_sustitucion(texto_cifrado, CLAVE)

print("="*80)
print("PROCESO DE DESCIFRADO - SUSTITUCIÓN")
print("="*80)
print(f"\nMensaje cifrado: {texto_cifrado}")
print(f"Clave:           {CLAVE}")

print("\nTabla de sustitución inversa:")
print(f"Cifrado: {CLAVE}")
print(f"         {'↓ '*26}")
print(f"Normal:  {ALFABETO_NORMAL}")

print("\n" + "-"*80)
print(f"{'Pos':<5} {'Cifrado':<10} {'Val':<5} {'→':<3} {'Original':<10} {'Val':<5}")
print("-"*80)

for d in detalles_desc:
    print(f"{d['pos']:<5} {d['cifrado']:<10} {d['val_cifr']:<5} {'→':<3} "
          f"{d['original']:<10} {d['val_orig']:<5}")

print("\n" + "="*80)
print(f"🔓 MENSAJE DESCIFRADO: {texto_descifrado}")
print("="*80)

# Verificación
print(f"\n✅ Verificación: {limpiar_texto(MENSAJE)} == {texto_descifrado} → {limpiar_texto(MENSAJE) == texto_descifrado}")
````

### 6. Análisis de Frecuencias

````python
def analisis_frecuencias(texto):
    """
    Realiza un análisis de frecuencias del texto.
    
    El análisis de frecuencias es la principal vulnerabilidad del
    cifrado de sustitución. Aunque la clave sea aleatoria, las
    frecuencias relativas de las letras se mantienen.
    
    En inglés, las letras más comunes son:
    E (12.7%), T (9.1%), A (8.2%), O (7.5%), I (7.0%), N (6.7%)
    
    Si un texto cifrado tiene mucha 'Q', probablemente Q representa E.
    
    Este análisis muestra:
    - Frecuencia absoluta de cada letra
    - Porcentaje relativo
    - Visualización con barras
    """
    texto_limpio = limpiar_texto(texto)
    contador = Counter(texto_limpio)
    total = len(texto_limpio)
    
    print("="*70)
    print("ANÁLISIS DE FRECUENCIAS")
    print("="*70)
    print(f"Total de letras analizadas: {total}\n")
    
    # Top 10 letras más comunes
    mas_comunes = contador.most_common(10)
    
    print(f"{'Letra':<8} {'Frecuencia':<12} {'Porcentaje':<12} {'Barra':<30}")
    print("-"*70)
    
    for letra, freq in mas_comunes:
        porcentaje = (freq / total) * 100
        barra = '█' * int(porcentaje * 2)  # Escala x2 para mejor visualización
        print(f"{letra:<8} {freq:<12} {porcentaje:>6.2f}%     {barra}")
    
    print("\n" + "="*70)
    print("💡 Frecuencias comunes en inglés:")
    print("   E (12.7%), T (9.1%), A (8.2%), O (7.5%), I (7.0%), N (6.7%)")
    print("\n💡 Un atacante puede usar estas frecuencias para romper el cifrado")

# Analizar el texto cifrado
print("\nAnálisis del mensaje original:")
analisis_frecuencias(MENSAJE)

print("\n" + "="*70)
print("\nAnálisis del mensaje cifrado:")
analisis_frecuencias(texto_cifrado)
````

# 📚 PARTE TEÓRICA

## 1. ¿Qué es el Cifrado de Sustitución?

El **Cifrado de Sustitución** (también llamado cifrado monoalfabético simple) es un método de cifrado donde cada letra del alfabeto se reemplaza por otra letra según una tabla fija.

### Características Principales

1. **Monoalfabético**: Cada letra siempre se sustituye por la misma letra
2. **Permutación**: La clave es una permutación completa del alfabeto
3. **Espacio de claves enorme**: 26! ≈ 4×10²⁶ posibilidades
4. **Vulnerable**: A pesar del gran espacio de claves, es vulnerable al análisis de frecuencias

### Diferencia con César

| Característica | César | Sustitución |
|----------------|-------|-------------|
| **Clave** | Un número (0-25) | Permutación del alfabeto |
| **Espacio de claves** | 26 | 26! ≈ 4×10²⁶ |
| **Patrón** | Desplazamiento fijo | Cualquier permutación |
| **Ejemplo** | A→D, B→E, C→F | A→Q, B→W, C→E |

### Historia

- Usado desde la **antigua Roma**
- **Julio César** usó una variante (César es un caso especial)
- Muy popular en la **época medieval**
- Fácil de usar manualmente (con tabla de sustitución)
- Roto definitivamente con el desarrollo del **análisis de frecuencias** en el siglo IX

## 2. Espacio de Claves - ¿Por qué 26!?

### Cálculo del Espacio de Claves

El número de permutaciones posibles del alfabeto se calcula así:

```
26! = 26 × 25 × 24 × ... × 2 × 1
```

**Razonamiento:**
- Para la primera letra (A), tenemos **26 opciones**
- Para la segunda letra (B), tenemos **25 opciones** (una ya usada)
- Para la tercera letra (C), tenemos **24 opciones**
- Y así sucesivamente...

**Total:** 26 × 25 × 24 × ... × 2 × 1 = **26!**

````python
import math

# Calcular 26!
factorial_26 = math.factorial(26)

print("Cálculo del Espacio de Claves")
print("="*60)
print(f"\n26! = {factorial_26:,}")
print(f"\nEn notación científica: {factorial_26:.2e}")
print(f"\nEsto es aproximadamente: 4 × 10²⁶")

print("\n" + "="*60)
print("Comparación con otros cifrados:")
print("="*60)
print(f"César:              26 claves")
print(f"Afín:               312 claves")
print(f"Sustitución:        {factorial_26:,} claves")
print(f"\nVigenère (5 letras): {26**5:,} claves")

print("\n💡 A pesar de tener más claves que Vigenère,")
print("   Sustitución es MÁS VULNERABLE por análisis de frecuencias")
````

## 3. Análisis de Frecuencias - La Gran Vulnerabilidad

### ¿Qué es el Análisis de Frecuencias?

El **análisis de frecuencias** es un método para romper cifrados de sustitución basado en el hecho de que en cualquier idioma, ciertas letras aparecen más frecuentemente que otras.

### Frecuencias en Inglés

| Letra | Frecuencia | Letra | Frecuencia |
|-------|-----------|-------|------------|
| E | 12.7% | N | 6.7% |
| T | 9.1% | R | 6.0% |
| A | 8.2% | S | 6.3% |
| O | 7.5% | H | 6.1% |
| I | 7.0% | D | 4.3% |

### ¿Cómo funciona el ataque?

1. **Contar frecuencias** en el texto cifrado
2. **Identificar** la letra más común (probablemente es 'E')
3. **Buscar patrones** comunes:
   - Palabras de 1 letra: A, I
   - Palabras de 2 letras: OF, TO, IN
   - Palabras de 3 letras: THE, AND
4. **Hacer suposiciones** basadas en el contexto
5. **Refinar** iterativamente hasta descifrar todo

### Ejemplo Práctico

````python
# Texto más largo para análisis
texto_largo = """
THE QUICK BROWN FOX JUMPS OVER THE LAZY DOG.
THIS SENTENCE CONTAINS EVERY LETTER OF THE ALPHABET.
FREQUENCY ANALYSIS IS THE MAIN WEAKNESS OF SUBSTITUTION CIPHERS.
THE LETTER E IS THE MOST COMMON LETTER IN ENGLISH TEXT.
"""

# Generar clave y cifrar
clave_ejemplo = generar_clave_aleatoria()
texto_largo_cifrado, _, _ = cifrar_sustitucion(texto_largo, clave_ejemplo)

print("Demostración del Análisis de Frecuencias")
print("="*70)
print(f"\nClave usada: {clave_ejemplo}")
print(f"\nTexto original ({len(limpiar_texto(texto_largo))} letras):")
print(limpiar_texto(texto_largo)[:100] + "...")
print(f"\nTexto cifrado:")
print(texto_largo_cifrado[:100] + "...")

# Analizar ambos
print("\n" + "="*70)
print("Análisis del texto ORIGINAL:")
analisis_frecuencias(texto_largo)

print("\n" + "="*70)
print("Análisis del texto CIFRADO:")
analisis_frecuencias(texto_largo_cifrado)

print("\n" + "="*70)
print("💡 Observa:")
print("   • Las frecuencias relativas se mantienen")
print("   • La letra más común en el cifrado probablemente representa 'E'")
print("   • Un atacante puede usar esto para deducir el texto original")
````

## 4. Comparación: César, Afín, Vigenère y Sustitución

### Tabla Comparativa

| Característica | César | Afín | Vigenère | Sustitución |
|----------------|-------|------|----------|-------------|
| **Tipo** | Monoalfabético | Monoalfabético | Polialfabético | Monoalfabético |
| **Clave** | 1 número | 2 números | Palabra | Permutación |
| **Espacio claves** | 26 | 312 | 26^n | 26! |
| **Análisis frecuencias** | Muy vulnerable | Muy vulnerable | Algo resistente | Muy vulnerable |
| **Seguridad** | Muy baja | Baja | Baja | Baja |
| **Complejidad** | Muy simple | Simple | Moderada | Simple |

### Cifrado del mismo mensaje

````python
mensaje_comp = "ATTACKATDAWN"

# César
def cesar_simple(texto, shift):
    resultado = ""
    for c in limpiar_texto(texto):
        resultado += chr((ord(c) - ord('A') + shift) % 26 + ord('A'))
    return resultado

# Resultados
cesar_res = cesar_simple(mensaje_comp, 3)
sust_res, _, _ = cifrar_sustitucion(mensaje_comp, CLAVE)

print("Comparación de Cifrados")
print("="*70)
print(f"Mensaje original:     {mensaje_comp}")
print(f"\nCésar (shift=3):      {cesar_res}")
print(f"  Patrón: Desplazamiento fijo de 3")
print(f"  A→D, T→W, T→W, A→D, etc.")
print(f"\nSustitución:          {sust_res}")
print(f"  Patrón: Según tabla de permutación")
print(f"  Clave: {CLAVE}")
print(f"  A→Q, T→Z, T→Z, A→Q, etc.")

print("\n" + "="*70)
print("Análisis de la letra 'A':")
print("-"*70)
print(f"En César:      A siempre → {cesar_simple('A', 3)}")
print(f"En Sustitución: A siempre → {cifrar_sustitucion('A', CLAVE)[0]}")
print(f"\n💡 Ambos son monoalfabéticos (misma letra → mismo cifrado)")
print(f"💡 César es un caso especial de Sustitución (permutación con patrón)")
````

## 5. Seguridad y Vulnerabilidades

### Fortalezas

1. **Espacio de claves gigante**: 26! hace que la fuerza bruta sea impráctica
2. **Fácil de implementar**: Usar manualmente con una tabla
3. **Rápido**: No requiere cálculos complejos

### Vulnerabilidades

1. **Análisis de frecuencias**: La principal debilidad
   - Frecuencias de letras se preservan
   - Patrones de palabras comunes son detectables
   - Con suficiente texto, se puede romper fácilmente

2. **Análisis de bigramas**: Pares de letras comunes
   - TH, HE, IN, ER, AN, RE son muy comunes
   - Ayudan a confirmar hipótesis

3. **Análisis de palabras**: Palabras cortas
   - Palabras de 1 letra: A, I
   - Palabras de 2 letras: OF, TO, IN, IT, IS
   - Palabras de 3 letras: THE, AND, FOR

### Herramientas Modernas

Un cifrado de sustitución se puede romper en segundos con:
- Software de análisis de frecuencias
- Algoritmos de optimización (hill climbing)
- Diccionarios de palabras comunes

### Conclusión

El cifrado de sustitución es:
- ✓ Excelente para **aprender** criptografía
- ✓ Interesante desde el punto de vista **histórico**
- ✗ **NO seguro** para comunicaciones reales
- ✗ **Fácilmente rompible** con análisis de frecuencias

### Uso Moderno

**NO usar** para proteger información real. Los algoritmos modernos son exponencialmente más seguros:
- **AES**: Estándar actual de cifrado simétrico
- **RSA**: Cifrado asimétrico
- **ChaCha20**: Cifrado de flujo moderno

## 📊 Suite de Pruebas Completa

````python
def suite_pruebas_sustitucion():
    """Suite completa de pruebas para validar la implementación."""
    
    print("="*80)
    print(" "*28 + "SUITE DE PRUEBAS")
    print("="*80)
    
    casos_prueba = [
        {"mensaje": "HELLO", "clave": "QWERTYUIOPASDFGHJKLZXCVBNM"},
        {"mensaje": "CRYPTOGRAPHY", "clave": "ZYXWVUTSRQPONMLKJIHGFEDCBA"},
        {"mensaje": "SUBSTITUTION", "clave": generar_clave_aleatoria()},
        {"mensaje": "SECURITY", "clave": generar_clave_aleatoria()},
        {"mensaje": "FREQUENCY ANALYSIS", "clave": generar_clave_aleatoria()},
    ]
    
    exitos = 0
    fallos = 0
    
    for i, caso in enumerate(casos_prueba, 1):
        print(f"\nPrueba {i}: {caso['mensaje']}")
        print("-"*80)
        
        try:
            # Cifrar
            cifrado, _, _ = cifrar_sustitucion(caso['mensaje'], caso['clave'])
            
            # Descifrar
            descifrado, _, _ = descifrar_sustitucion(cifrado, caso['clave'])
            
            # Verificar
            original_limpio = limpiar_texto(caso['mensaje'])
            descifrado_correcto = (descifrado == original_limpio)
            
            print(f"Mensaje:     {caso['mensaje']}")
            print(f"Limpio:      {original_limpio}")
            print(f"Clave:       {caso['clave']}")
            print(f"Cifrado:     {cifrado}")
            print(f"Descifrado:  {descifrado}")
            
            if descifrado_correcto:
                print("✅ ÉXITO")
                exitos += 1
            else:
                print("❌ FALLO")
                print(f"   Descifrado incorrecto: {descifrado} != {original_limpio}")
                fallos += 1
        
        except Exception as e:
            print(f"❌ ERROR: {e}")
            fallos += 1
    
    # Resumen
    print("\n" + "="*80)
    print(" "*33 + "RESUMEN")
    print("="*80)
    print(f"Total de pruebas: {len(casos_prueba)}")
    print(f"✅ Éxitos: {exitos}")
    print(f"❌ Fallos: {fallos}")
    print(f"Tasa de éxito: {(exitos/len(casos_prueba)*100):.1f}%")
    print("="*80)

# Ejecutar suite de pruebas
suite_pruebas_sustitucion()
````

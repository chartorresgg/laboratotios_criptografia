# 📝 PARTE PRÁCTICA

## 🎯 Ejemplo Completo: Cifrar y Descifrar "HELP"

### ⚙️ Configuración del Ejemplo

Modifica estos valores para experimentar:

````python
import numpy as np
from math import gcd

# ============================================
# CONFIGURACIÓN - Modifica aquí
# ============================================

MENSAJE = "HELP"              # Mensaje a cifrar

# Matriz clave 2×2
# IMPORTANTE: El determinante debe ser coprimo con 26
MATRIZ_CLAVE = np.array([
    [3, 3],
    [2, 5]
])

ALFABETO = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'

print(f"📝 Mensaje: {MENSAJE}")
print(f"\n🔑 Matriz clave:")
print(MATRIZ_CLAVE)
print(f"\nTamaño de bloque: {len(MATRIZ_CLAVE)}")
````

### 1. Funciones Auxiliares: MCD y Algoritmo de Euclides

````python
def gcd_extendido(a, b):
    """
    Algoritmo de Euclides Extendido.
    
    Encuentra números x, y tales que: a*x + b*y = MCD(a, b)
    Esto es esencial para calcular el inverso modular.
    
    Proceso recursivo:
    1. Caso base: si a = 0, entonces MCD = b, x = 0, y = 1
    2. Caso recursivo: calcular MCD(b mod a, a)
    3. Ajustar x, y basándose en la recursión
    
    Retorna:
        Tupla (mcd, x, y)
    """
    if a == 0:
        return b, 0, 1
    
    mcd, x1, y1 = gcd_extendido(b % a, a)
    
    x = y1 - (b // a) * x1
    y = x1
    
    return mcd, x, y

def inverso_modular(a, m):
    """
    Calcula el inverso modular de 'a' módulo 'm'.
    
    El inverso modular a⁻¹ cumple: a × a⁻¹ ≡ 1 (mod m)
    Solo existe si MCD(a, m) = 1
    
    Parámetros:
        a: número a invertir
        m: módulo (26 para nuestro alfabeto)
    
    Retorna:
        Inverso modular si existe, None si no
    """
    mcd, x, _ = gcd_extendido(a, m)
    
    if mcd != 1:
        return None  # No existe inverso
    
    # Asegurar resultado positivo
    return (x % m + m) % m

# Pruebas
print("Ejemplos de inversos modulares (mod 26):")
print("="*50)
for num in [1, 3, 5, 7, 9, 11, 15]:
    inv = inverso_modular(num, 26)
    verificacion = (num * inv) % 26 if inv else 0
    print(f"{num:2d}⁻¹ = {inv:2d}    Verificación: {num} × {inv} ≡ {verificacion} (mod 26)")
````

### 2. Función: Calcular Determinante

````python
def determinante_2x2(matriz):
    """
    Calcula el determinante de una matriz 2×2.
    
    Fórmula para matriz [[a, b], [c, d]]:
    det = a*d - b*c
    
    Ejemplo:
        [[3, 3],     det = 3*5 - 3*2 = 15 - 6 = 9
         [2, 5]]
    """
    return int(matriz[0][0] * matriz[1][1] - matriz[0][1] * matriz[1][0])

def determinante_3x3(matriz):
    """
    Calcula el determinante de una matriz 3×3.
    
    Usa la regla de Sarrus:
    Para matriz:
        [[a, b, c],
         [d, e, f],
         [g, h, i]]
    
    det = (a*e*i + b*f*g + c*d*h) - (c*e*g + b*d*i + a*f*h)
    """
    a, b, c = matriz[0]
    d, e, f = matriz[1]
    g, h, i = matriz[2]
    
    det = (a*e*i + b*f*g + c*d*h) - (c*e*g + b*d*i + a*f*h)
    return int(det)

# Calcular determinante de nuestra matriz
det = determinante_2x2(MATRIZ_CLAVE)
det_mod = det % 26

print(f"Matriz:")
print(MATRIZ_CLAVE)
print(f"\nDeterminante = {MATRIZ_CLAVE[0,0]}×{MATRIZ_CLAVE[1,1]} - {MATRIZ_CLAVE[0,1]}×{MATRIZ_CLAVE[1,0]}")
print(f"             = {MATRIZ_CLAVE[0,0]*MATRIZ_CLAVE[1,1]} - {MATRIZ_CLAVE[0,1]*MATRIZ_CLAVE[1,0]}")
print(f"             = {det}")
print(f"\nDeterminante mod 26 = {det_mod}")
print(f"MCD({det_mod}, 26) = {gcd(det_mod, 26)}")

if gcd(det_mod, 26) == 1:
    print(f"\n✓ La matriz es válida (determinante coprimo con 26)")
else:
    print(f"\n✗ La matriz NO es válida (determinante no coprimo con 26)")
````

### 3. Función: Matriz Inversa Modular

````python
def matriz_cofactores_2x2(matriz):
    """
    Calcula la matriz de cofactores de una matriz 2×2.
    
    Para matriz [[a, b], [c, d]], los cofactores son:
    [[d, -b], [-c, a]]
    
    Esto es parte del proceso para calcular la inversa.
    """
    return np.array([
        [matriz[1][1], -matriz[0][1]],
        [-matriz[1][0], matriz[0][0]]
    ])

def matriz_inversa_modular(matriz, modulo=26):
    """
    Calcula la matriz inversa modular.
    
    Proceso completo para matriz 2×2:
    1. Calcular determinante: det
    2. Verificar que MCD(det, 26) = 1
    3. Calcular inverso modular del determinante: det⁻¹
    4. Calcular matriz de cofactores
    5. Transponer matriz de cofactores (= adjunta)
    6. Matriz inversa = det⁻¹ × adjunta (mod 26)
    
    Parámetros:
        matriz: matriz cuadrada (numpy array)
        modulo: módulo a usar (default 26)
    
    Retorna:
        Matriz inversa modular
    """
    n = len(matriz)
    
    # Paso 1: Calcular determinante
    if n == 2:
        det = determinante_2x2(matriz)
    elif n == 3:
        det = determinante_3x3(matriz)
    else:
        raise ValueError("Solo matrices 2×2 o 3×3 soportadas")
    
    # Determinante módulo 26
    det_mod = det % modulo
    
    # Paso 2: Verificar coprimalidad
    if gcd(det_mod, modulo) != 1:
        raise ValueError(f"Determinante {det_mod} no es coprimo con {modulo}")
    
    # Paso 3: Inverso modular del determinante
    det_inv = inverso_modular(det_mod, modulo)
    
    # Paso 4: Matriz de cofactores
    if n == 2:
        cofactores = matriz_cofactores_2x2(matriz)
    else:
        # Para 3×3 (implementación simplificada)
        raise NotImplementedError("Matriz 3×3 no implementada en esta celda")
    
    # Paso 5: Adjunta (transpuesta de cofactores)
    adjunta = cofactores.T
    
    # Paso 6: Matriz inversa
    inversa = (det_inv * adjunta) % modulo
    
    return inversa.astype(int)

# Calcular matriz inversa
matriz_inversa = matriz_inversa_modular(MATRIZ_CLAVE)

print("CÁLCULO DE LA MATRIZ INVERSA")
print("="*60)
print(f"\nMatriz original K:")
print(MATRIZ_CLAVE)

print(f"\nPaso 1: det(K) = {det} ≡ {det_mod} (mod 26)")

det_inv = inverso_modular(det_mod, 26)
print(f"Paso 2: det⁻¹ = {det_mod}⁻¹ = {det_inv} (mod 26)")
print(f"        Verificación: {det_mod} × {det_inv} ≡ {(det_mod * det_inv) % 26} (mod 26) ✓")

cofactores = matriz_cofactores_2x2(MATRIZ_CLAVE)
print(f"\nPaso 3: Matriz de cofactores:")
print(cofactores)

adjunta = cofactores.T
print(f"\nPaso 4: Adjunta (transpuesta):")
print(adjunta)

print(f"\nPaso 5: K⁻¹ = {det_inv} × adjunta (mod 26)")
print(f"\nMatriz inversa K⁻¹:")
print(matriz_inversa)

# Verificar que K × K⁻¹ = I (mod 26)
verificacion = (MATRIZ_CLAVE @ matriz_inversa) % 26
print(f"\n✓ Verificación: K × K⁻¹ (mod 26) =")
print(verificacion)
print("(Debe ser la matriz identidad)")
````

### 4. Funciones de Conversión

````python
def texto_a_numeros(texto):
    """
    Convierte texto a lista de números (A=0, B=1, ..., Z=25).
    
    Proceso:
    1. Limpiar texto (solo letras, mayúsculas)
    2. Para cada letra: ord(letra) - ord('A')
    
    Ejemplo:
        "HELP" → [7, 4, 11, 15]
        H=7, E=4, L=11, P=15
    """
    texto_limpio = ''.join(c.upper() for c in texto if c.isalpha())
    return [ord(c) - ord('A') for c in texto_limpio]

def numeros_a_texto(numeros):
    """
    Convierte lista de números a texto (0=A, 1=B, ..., 25=Z).
    
    Proceso:
    1. Para cada número: chr(numero + ord('A'))
    2. Unir todos los caracteres
    
    Ejemplo:
        [7, 4, 11, 15] → "HELP"
    """
    return ''.join(chr(n + ord('A')) for n in numeros)

# Convertir mensaje
numeros_mensaje = texto_a_numeros(MENSAJE)

print(f"Mensaje: {MENSAJE}")
print(f"Números: {numeros_mensaje}")
print(f"\nDetalle:")
for letra, num in zip(MENSAJE, numeros_mensaje):
    print(f"  {letra} → {num:2d}")
````

### 5. Función: Cifrar

````python
def cifrar_hill(texto, matriz_clave):
    """
    Cifra un texto usando el Cifrado de Hill.
    
    Proceso completo:
    1. Convertir texto a números
    2. Dividir en bloques del tamaño de la matriz
    3. Si es necesario, agregar padding (letra 'X' = 23)
    4. Para cada bloque:
       a. Crear vector columna con el bloque
       b. Multiplicar: C = K × M (mod 26)
       c. Guardar resultado
    5. Unir todos los bloques cifrados
    
    Fórmula matemática para matriz 2×2:
        [c1]   [k11 k12]   [m1]
        [c2] = [k21 k22] × [m2] (mod 26)
    
    Expandido:
        c1 = (k11×m1 + k12×m2) mod 26
        c2 = (k21×m1 + k22×m2) mod 26
    
    Parámetros:
        texto: mensaje a cifrar
        matriz_clave: matriz cuadrada (numpy array)
    
    Retorna:
        Tupla (texto_cifrado, lista_de_bloques_detallados)
    """
    n = len(matriz_clave)  # Tamaño de bloque
    
    # Convertir a números
    numeros = texto_a_numeros(texto)
    
    # Agregar padding si es necesario
    while len(numeros) % n != 0:
        numeros.append(23)  # 'X'
    
    # Cifrar por bloques
    texto_cifrado = []
    bloques_detallados = []
    
    for i in range(0, len(numeros), n):
        # Extraer bloque
        bloque = numeros[i:i+n]
        vector = np.array(bloque).reshape(n, 1)
        
        # Multiplicar matriz × vector (mod 26)
        resultado = (matriz_clave @ vector) % 26
        
        # Guardar
        bloque_cifrado = resultado.flatten().tolist()
        texto_cifrado.extend(bloque_cifrado)
        
        # Detalles
        bloques_detallados.append({
            'indice': i // n,
            'bloque_original': bloque,
            'texto_original': numeros_a_texto(bloque),
            'bloque_cifrado': bloque_cifrado,
            'texto_cifrado': numeros_a_texto(bloque_cifrado)
        })
    
    return numeros_a_texto(texto_cifrado), bloques_detallados

# CIFRAR EL MENSAJE
texto_cifrado, bloques_cifrado = cifrar_hill(MENSAJE, MATRIZ_CLAVE)

print("="*80)
print("PROCESO DE CIFRADO - HILL")
print("="*80)
print(f"\nMensaje original: {MENSAJE}")
print(f"En números: {numeros_mensaje}")
print(f"\nMatriz clave K:")
print(MATRIZ_CLAVE)

print(f"\nFórmula: C = K × M (mod 26)")
print("\nDonde:")
print("  K = matriz clave")
print("  M = vector del mensaje")
print("  C = vector cifrado")

# Mostrar detalles de cada bloque
for bloque in bloques_cifrado:
    print(f"\n" + "─"*80)
    print(f"Bloque {bloque['indice'] + 1}:")
    print("─"*80)
    print(f"Mensaje: {bloque['texto_original']} → {bloque['bloque_original']}")
    
    # Mostrar operación matricial
    m = bloque['bloque_original']
    c = bloque['bloque_cifrado']
    K = MATRIZ_CLAVE
    
    print(f"\nOperación:")
    for i in range(len(K)):
        print(f"│ {K[i][0]:2d} {K[i][1]:2d} │   │ {m[i]:2d} │", end="")
        if i == len(K) // 2:
            print(f"   =   │ {c[i]:2d} │", end="")
        else:
            print(f"       │ {c[i]:2d} │", end="")
        print(" (mod 26)")
    
    # Cálculos detallados
    print(f"\nCálculos:")
    for i in range(len(K)):
        operacion = " + ".join([f"{K[i][j]}×{m[j]}" for j in range(len(m))])
        suma = sum(K[i][j] * m[j] for j in range(len(m)))
        print(f"  c{i+1} = {operacion} = {suma} ≡ {c[i]} (mod 26)")
    
    print(f"\nCifrado: {bloque['texto_cifrado']} → {bloque['bloque_cifrado']}")

print("\n" + "="*80)
print(f"🔒 MENSAJE CIFRADO: {texto_cifrado}")
print("="*80)
````

### 6. Función: Descifrar

````python
def descifrar_hill(texto_cifrado, matriz_clave):
    """
    Descifra un texto usando el Cifrado de Hill.
    
    Proceso completo:
    1. Calcular la matriz inversa de la clave: K⁻¹
    2. Convertir texto cifrado a números
    3. Dividir en bloques
    4. Para cada bloque:
       a. Crear vector columna con el bloque cifrado
       b. Multiplicar: M = K⁻¹ × C (mod 26)
       c. Guardar resultado
    5. Unir todos los bloques descifrados
    
    Fórmula matemática:
        [m1]   [k⁻¹₁₁ k⁻¹₁₂]   [c1]
        [m2] = [k⁻¹₂₁ k⁻¹₂₂] × [c2] (mod 26)
    
    Parámetros:
        texto_cifrado: mensaje cifrado
        matriz_clave: matriz usada en el cifrado
    
    Retorna:
        Tupla (texto_descifrado, lista_de_bloques_detallados)
    """
    # Calcular matriz inversa
    matriz_inv = matriz_inversa_modular(matriz_clave)
    n = len(matriz_clave)
    
    # Convertir a números
    numeros = texto_a_numeros(texto_cifrado)
    
    # Descifrar por bloques
    texto_descifrado = []
    bloques_detallados = []
    
    for i in range(0, len(numeros), n):
        # Extraer bloque
        bloque = numeros[i:i+n]
        vector = np.array(bloque).reshape(n, 1)
        
        # Multiplicar matriz_inversa × vector (mod 26)
        resultado = (matriz_inv @ vector) % 26
        
        # Guardar
        bloque_descifrado = resultado.flatten().tolist()
        texto_descifrado.extend(bloque_descifrado)
        
        # Detalles
        bloques_detallados.append({
            'indice': i // n,
            'bloque_cifrado': bloque,
            'texto_cifrado': numeros_a_texto(bloque),
            'bloque_descifrado': bloque_descifrado,
            'texto_descifrado': numeros_a_texto(bloque_descifrado)
        })
    
    return numeros_a_texto(texto_descifrado), bloques_detallados

# DESCIFRAR EL MENSAJE
texto_descifrado, bloques_descifrado = descifrar_hill(texto_cifrado, MATRIZ_CLAVE)

print("="*80)
print("PROCESO DE DESCIFRADO - HILL")
print("="*80)
print(f"\nMensaje cifrado: {texto_cifrado}")
print(f"\nMatriz clave K:")
print(MATRIZ_CLAVE)
print(f"\nMatriz inversa K⁻¹:")
print(matriz_inversa)

print(f"\nFórmula: M = K⁻¹ × C (mod 26)")

# Mostrar detalles de cada bloque
for bloque in bloques_descifrado:
    print(f"\n" + "─"*80)
    print(f"Bloque {bloque['indice'] + 1}:")
    print("─"*80)
    print(f"Cifrado: {bloque['texto_cifrado']} → {bloque['bloque_cifrado']}")
    
    # Mostrar operación matricial
    c = bloque['bloque_cifrado']
    m = bloque['bloque_descifrado']
    K_inv = matriz_inversa
    
    print(f"\nOperación:")
    for i in range(len(K_inv)):
        print(f"│ {K_inv[i][0]:2d} {K_inv[i][1]:2d} │   │ {c[i]:2d} │", end="")
        if i == len(K_inv) // 2:
            print(f"   =   │ {m[i]:2d} │", end="")
        else:
            print(f"       │ {m[i]:2d} │", end="")
        print(" (mod 26)")
    
    # Cálculos detallados
    print(f"\nCálculos:")
    for i in range(len(K_inv)):
        operacion = " + ".join([f"{K_inv[i][j]}×{c[j]}" for j in range(len(c))])
        suma = sum(K_inv[i][j] * c[j] for j in range(len(c)))
        print(f"  m{i+1} = {operacion} = {suma} ≡ {m[i]} (mod 26)")
    
    print(f"\nDescifrado: {bloque['texto_descifrado']} → {bloque['bloque_descifrado']}")

print("\n" + "="*80)
print(f"🔓 MENSAJE DESCIFRADO: {texto_descifrado}")
print("="*80)

# Verificación
print(f"\n✅ Verificación: {MENSAJE} == {texto_descifrado} → {MENSAJE == texto_descifrado}")
````

````python
print("="*80)
print(" "*30 + "RESUMEN")
print("="*80)
print(f"\nMensaje original:    {MENSAJE}")
print(f"\nMatriz clave:")
print(MATRIZ_CLAVE)
print(f"\nMatriz inversa:")
print(matriz_inversa)
print(f"\nMensaje cifrado:     {texto_cifrado}")
print(f"Mensaje descifrado:  {texto_descifrado}")
print(f"\n✓ Cifrado y descifrado exitosos!")
print("\n💡 Hill cifra bloques de letras, no letra por letra")
print(f"   Tamaño de bloque: {len(MATRIZ_CLAVE)} letras")
print("="*80)
````

# 📚 PARTE TEÓRICA

## 1. ¿Qué es el Cifrado de Hill?

El **Cifrado de Hill** fue inventado por **Lester S. Hill** en 1929. Es el primer cifrado por bloques basado en álgebra lineal.

### Características Principales

1. **Cifrado por bloques**: Procesa múltiples letras a la vez
2. **Usa matrices**: La clave es una matriz cuadrada
3. **Álgebra lineal**: Multiplicación de matrices y módulo 26
4. **Más seguro**: Resiste análisis de frecuencias simple

### Fórmulas Matemáticas

**Cifrado:**
```
C = K × M (mod 26)
```

**Descifrado:**
```
M = K⁻¹ × C (mod 26)
```

Donde:
- `K`: Matriz clave (n×n)
- `M`: Vector del mensaje (n×1)
- `C`: Vector cifrado (n×1)
- `K⁻¹`: Matriz inversa de K (módulo 26)

### Condición Importante

**MCD(det(K), 26) = 1**

El determinante de la matriz clave debe ser **coprimo con 26**. Esto garantiza que exista la matriz inversa módulo 26.

### Historia

- **1929**: Lester S. Hill publica el cifrado
- **Primera Guerra Mundial**: Utilizado por el ejército
- **Innovación**: Primer uso práctico de álgebra lineal en criptografía
- **Actualidad**: Base conceptual para cifrados modernos por bloques

## 2. Conceptos de Álgebra Lineal

### 2.1 Matrices

Una **matriz** es un arreglo rectangular de números:

```
Matriz 2×2:        Matriz 3×3:
[a b]              [a b c]
[c d]              [d e f]
                   [g h i]
```

### 2.2 Multiplicación de Matrices

Para multiplicar una matriz 2×2 por un vector 2×1:

```
[a b]   [m₁]   [a×m₁ + b×m₂]
[c d] × [m₂] = [c×m₁ + d×m₂]
```

**Ejemplo:**
```
[3 3]   [7]    [3×7 + 3×4]   [33]   [7]
[2 5] × [4] =  [2×7 + 5×4] = [34] = [8] (mod 26)
```

### 2.3 Determinante

El **determinante** es un número asociado a una matriz cuadrada.

**Para matriz 2×2:**
```
det([a b]) = a×d - b×c
    [c d]
```

**Ejemplo:**
```
det([3 3]) = 3×5 - 3×2 = 15 - 6 = 9
    [2 5]
```

### 2.4 Matriz Inversa

La **matriz inversa** K⁻¹ cumple:
```
K × K⁻¹ = I
```

Donde I es la matriz identidad:
```
[1 0]
[0 1]
```

**Para calcular K⁻¹ (mod 26):**
1. Calcular det(K)
2. Verificar que MCD(det(K), 26) = 1
3. Calcular det(K)⁻¹ (mod 26)
4. Calcular matriz adjunta
5. K⁻¹ = det⁻¹ × adj(K) (mod 26)

## 3. Ejemplo Paso a Paso Completo

````python
print("EJEMPLO COMPLETO: Cifrar 'HE' con matriz [[3,3],[2,5]]")
print("="*70)

# Datos
mensaje_ej = "HE"
matriz_ej = np.array([[3, 3], [2, 5]])

print(f"\nMensaje: {mensaje_ej}")
print(f"Matriz K:\n{matriz_ej}")

# Paso 1: Convertir a números
m1, m2 = ord('H') - ord('A'), ord('E') - ord('A')
print(f"\n**PASO 1: Conversión a números**")
print(f"H = {m1}, E = {m2}")
print(f"Vector M = [{m1}, {m2}]ᵀ")

# Paso 2: Multiplicar
print(f"\n**PASO 2: Multiplicación matricial**")
print(f"C = K × M (mod 26)")
print(f"\n[{matriz_ej[0,0]} {matriz_ej[0,1]}]   [{m1}]   [c₁]")
print(f"[{matriz_ej[1,0]} {matriz_ej[1,1]}] × [{m2}] = [c₂]")

c1_sin_mod = matriz_ej[0,0]*m1 + matriz_ej[0,1]*m2
c2_sin_mod = matriz_ej[1,0]*m1 + matriz_ej[1,1]*m2

print(f"\nc₁ = {matriz_ej[0,0]}×{m1} + {matriz_ej[0,1]}×{m2} = {matriz_ej[0,0]*m1} + {matriz_ej[0,1]*m2} = {c1_sin_mod}")
print(f"c₂ = {matriz_ej[1,0]}×{m1} + {matriz_ej[1,1]}×{m2} = {matriz_ej[1,0]*m1} + {matriz_ej[1,1]*m2} = {c2_sin_mod}")

c1 = c1_sin_mod % 26
c2 = c2_sin_mod % 26

print(f"\nAplicando módulo 26:")
print(f"c₁ = {c1_sin_mod} mod 26 = {c1}")
print(f"c₂ = {c2_sin_mod} mod 26 = {c2}")

letra_c1 = chr(c1 + ord('A'))
letra_c2 = chr(c2 + ord('A'))

print(f"\n**PASO 3: Conversión a letras**")
print(f"{c1} → {letra_c1}")
print(f"{c2} → {letra_c2}")

print(f"\n{'='*70}")
print(f"RESULTADO: '{mensaje_ej}' → '{letra_c1}{letra_c2}'")
print(f"{'='*70}")
````

## 4. ¿Por qué det(K) debe ser coprimo con 26?

### Explicación Matemática

Para descifrar, necesitamos la **matriz inversa** K⁻¹.

La matriz inversa módulo 26 solo existe si:
```
MCD(det(K), 26) = 1
```

### ¿Por qué?

1. Para calcular K⁻¹, necesitamos det(K)⁻¹ (mod 26)
2. El inverso modular solo existe si son coprimos
3. Si no son coprimos, no podemos descifrar

### Factores de 26

```
26 = 2 × 13
```

Por lo tanto, det(K) **NO puede ser**:
- Múltiplo de 2 (números pares)
- Múltiplo de 13

### Valores Válidos de det(K) mod 26

Los valores que son coprimos con 26:

````python
print("Determinantes válidos (coprimos con 26):")
print("="*60)
validos = [d for d in range(26) if gcd(d, 26) == 1]
print(f"\nTotal: {len(validos)} valores")
print(f"\nValores: {validos}")

print("\n" + "="*60)
print("Ejemplos de matrices VÁLIDAS e INVÁLIDAS:")
print("="*60)

ejemplos = [
    ([[3, 3], [2, 5]], "VÁLIDA"),
    ([[6, 24], [1, 13]], "VÁLIDA"),
    ([[2, 3], [4, 5]], "INVÁLIDA (det=2, par)"),
    ([[1, 2], [3, 4]], "INVÁLIDA (det=-2≡24, par)"),
    ([[1, 1], [13, 14]], "INVÁLIDA (det=1, pero 1 no es útil)"),
]

for matriz, estado in ejemplos:
    m = np.array(matriz)
    det = determinante_2x2(m)
    det_mod = det % 26
    mcd_val = gcd(det_mod, 26)
    print(f"\nMatriz: {matriz[0]}")
    print(f"        {matriz[1]}")
    print(f"det = {det} ≡ {det_mod} (mod 26)")
    print(f"MCD({det_mod}, 26) = {mcd_val}")
    print(f"Estado: {estado}")
````

## 5. Comparación con Otros Cifrados

### Tabla Comparativa

| Característica | César | Afín | Vigenère | Sustitución | **Hill** |
|----------------|-------|------|----------|-------------|----------|
| **Tipo** | Letra por letra | Letra por letra | Letra por letra | Letra por letra | **Por bloques** |
| **Clave** | 1 número | 2 números | Palabra | Permutación | **Matriz** |
| **Matemática** | Suma | Lineal | Suma repetida | Mapeo | **Álgebra lineal** |
| **Análisis frecuencias** | Muy vulnerable | Muy vulnerable | Algo resistente | Muy vulnerable | **Más resistente** |
| **Seguridad** | Muy baja | Baja | Baja | Baja | **Moderada** |
| **Complejidad** | Muy simple | Simple | Moderada | Simple | **Alta** |

### Ventajas de Hill

1. **Difusión**: Cada letra del bloque afecta a todas las cifradas
2. **Resiste frecuencias simples**: La misma letra puede cifrarse diferente según contexto
3. **Base teórica**: Fundamento de cifrados modernos por bloques
4. **Matemáticamente elegante**: Usa conceptos de álgebra lineal

### Desventajas

1. **Vulnerable a ataques conocidos**: Con texto plano conocido se puede romper
2. **Complejidad**: Requiere operaciones matriciales
3. **Padding**: Necesita agregar letras si el mensaje no es múltiplo del tamaño de bloque
4. **Generación de claves**: No todas las matrices sirven

## 6. Seguridad del Cifrado de Hill

### Fortalezas

1. **Difusión**: Cada letra influye en todo el bloque cifrado
2. **Espacio de claves grande**: Para matrices 2×2, hay muchas opciones válidas
3. **Resiste análisis simple**: No hay mapeo directo 1:1 entre letras

### Vulnerabilidades

1. **Ataque por texto conocido (Known Plaintext Attack)**:
   - Con n² pares (texto, cifrado) se puede recuperar la matriz
   - Para matriz 2×2: solo necesitas 2 bloques conocidos (4 letras)

2. **Linealidad**:
   - El cifrado es una transformación lineal
   - Esto facilita ataques algebraicos

3. **Frecuencias de bloques**:
   - Aunque las letras individuales están protegidas
   - Los bloques completos pueden repetirse

## 📊 Suite de Pruebas Completa

````python
def suite_pruebas_hill():
    """Suite completa de pruebas para validar la implementación."""
    
    print("="*80)
    print(" "*28 + "SUITE DE PRUEBAS")
    print("="*80)
    
    casos_prueba = [
        {"mensaje": "HELP", "matriz": np.array([[3, 3], [2, 5]])},
        {"mensaje": "ATTACK", "matriz": np.array([[6, 24], [1, 13]])},
        {"mensaje": "CODE", "matriz": np.array([[5, 8], [17, 3]])},
        {"mensaje": "HILL", "matriz": np.array([[9, 4], [5, 7]])},
        {"mensaje": "CRYPTOGRAPHY", "matriz": np.array([[11, 8], [3, 7]])},
    ]
    
    exitos = 0
    fallos = 0
    
    for i, caso in enumerate(casos_prueba, 1):
        print(f"\nPrueba {i}: {caso['mensaje']}")
        print("-"*80)
        
        try:
            # Validar matriz
            det = determinante_2x2(caso['matriz'])
            det_mod = det % 26
            
            print(f"Mensaje:     {caso['mensaje']}")
            print(f"Matriz:      {caso['matriz'][0].tolist()}")
            print(f"             {caso['matriz'][1].tolist()}")
            print(f"det(K) = {det} ≡ {det_mod} (mod 26)")
            print(f"MCD({det_mod}, 26) = {gcd(det_mod, 26)}")
            
            if gcd(det_mod, 26) != 1:
                print("❌ FALLO: Matriz no válida (det no coprimo con 26)")
                fallos += 1
                continue
            
            # Cifrar
            cifrado, _ = cifrar_hill(caso['mensaje'], caso['matriz'])
            
            # Descifrar
            descifrado, _ = descifrar_hill(cifrado, caso['matriz'])
            
            # Verificar (eliminar padding si existe)
            descifrado_limpio = descifrado.rstrip('X')
            correcto = caso['mensaje'].startswith(descifrado_limpio)
            
            print(f"Cifrado:     {cifrado}")
            print(f"Descifrado:  {descifrado}")
            
            if correcto:
                print("✅ ÉXITO")
                exitos += 1
            else:
                print("❌ FALLO")
                print(f"   Descifrado incorrecto: {descifrado} != {caso['mensaje']}")
                fallos += 1
        
        except Exception as e:
            print(f"❌ ERROR: {e}")
            fallos += 1
    
    # Resumen
    print("\n" + "="*80)
    print(" "*33 + "RESUMEN")
    print("="*80)
    print(f"Total de pruebas: {len(casos_prueba)}")
    print(f"✅ Éxitos: {exitos}")
    print(f"❌ Fallos: {fallos}")
    print(f"Tasa de éxito: {(exitos/len(casos_prueba)*100):.1f}%")
    print("="*80)

# Ejecutar suite de pruebas
suite_pruebas_hill()
````

## 🎓 Conclusiones y Referencias

### Lo que aprendimos

1. **Cifrado por bloques**: Procesa múltiples letras simultáneamente
2. **Álgebra lineal**: Multiplicación de matrices, determinantes, inversas modulares
3. **Condición de validez**: det(K) debe ser coprimo con 26
4. **Difusión**: Cada letra del bloque afecta a todo el cifrado
5. **Seguridad mejorada**: Más resistente que cifrados letra por letra

### Aplicaciones Históricas

- **1929**: Lester S. Hill inventa el cifrado
- **Años 30-40**: Uso militar limitado
- **Importancia teórica**: Primer cifrado algebraico práctico
- **Legado**: Base conceptual para DES, AES y cifrados modernos

### Evolución hacia Cifrados Modernos

El cifrado de Hill introdujo conceptos que son fundamentales hoy:

1. **Cifrado por bloques**: DES (1977), AES (2001)
2. **Difusión**: Principio de Shannon implementado en cifrados modernos
3. **Transformaciones algebraicas**: Base de la criptografía moderna

### Cifrados Modernos

- **AES** (Advanced Encryption Standard): Estándar actual
- **DES** (Data Encryption Standard): Predecesor de AES
- **Blowfish, Twofish**: Alternativas modernas

````python
import numpy as np
from math import gcd

# ============================================
# CONFIGURACIÓN - Modifica aquí
# ============================================

MENSAJE = "HI"              # Mensaje a cifrar

# Matriz clave 2×2
# IMPORTANTE: El determinante debe ser coprimo con 26
MATRIZ_CLAVE = np.array([
    [7, 1],
    [2, 1]
])

ALFABETO = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'

print(f"📝 Mensaje: {MENSAJE}")
print(f"\n🔑 Matriz clave:")
print(MATRIZ_CLAVE)
print(f"\nTamaño de bloque: {len(MATRIZ_CLAVE)}")
````

### 1. Funciones Auxiliares: MCD y Algoritmo de Euclides

````python
def gcd_extendido(a, b):
    """
    Algoritmo de Euclides Extendido.
    
    Encuentra números x, y tales que: a*x + b*y = MCD(a, b)
    Esto es esencial para calcular el inverso modular.
    
    Proceso recursivo:
    1. Caso base: si a = 0, entonces MCD = b, x = 0, y = 1
    2. Caso recursivo: calcular MCD(b mod a, a)
    3. Ajustar x, y basándose en la recursión
    
    Retorna:
        Tupla (mcd, x, y)
    """
    if a == 0:
        return b, 0, 1
    
    mcd, x1, y1 = gcd_extendido(b % a, a)
    
    x = y1 - (b // a) * x1
    y = x1
    
    return mcd, x, y

def inverso_modular(a, m):
    """
    Calcula el inverso modular de 'a' módulo 'm'.
    
    El inverso modular a⁻¹ cumple: a × a⁻¹ ≡ 1 (mod m)
    Solo existe si MCD(a, m) = 1
    
    Parámetros:
        a: número a invertir
        m: módulo (26 para nuestro alfabeto)
    
    Retorna:
        Inverso modular si existe, None si no
    """
    mcd, x, _ = gcd_extendido(a, m)
    
    if mcd != 1:
        return None  # No existe inverso
    
    # Asegurar resultado positivo
    return (x % m + m) % m

# Pruebas
print("Ejemplos de inversos modulares (mod 26):")
print("="*50)
for num in [1, 3, 5, 7, 9, 11, 15]:
    inv = inverso_modular(num, 26)
    verificacion = (num * inv) % 26 if inv else 0
    print(f"{num:2d}⁻¹ = {inv:2d}    Verificación: {num} × {inv} ≡ {verificacion} (mod 26)")
````

### 2. Función: Calcular Determinante

````python
def determinante_2x2(matriz):
    """
    Calcula el determinante de una matriz 2×2.
    
    Fórmula para matriz [[a, b], [c, d]]:
    det = a*d - b*c
    
    Ejemplo:
        [[3, 3],     det = 3*5 - 3*2 = 15 - 6 = 9
         [2, 5]]
    """
    return int(matriz[0][0] * matriz[1][1] - matriz[0][1] * matriz[1][0])

def determinante_3x3(matriz):
    """
    Calcula el determinante de una matriz 3×3.
    
    Usa la regla de Sarrus:
    Para matriz:
        [[a, b, c],
         [d, e, f],
         [g, h, i]]
    
    det = (a*e*i + b*f*g + c*d*h) - (c*e*g + b*d*i + a*f*h)
    """
    a, b, c = matriz[0]
    d, e, f = matriz[1]
    g, h, i = matriz[2]
    
    det = (a*e*i + b*f*g + c*d*h) - (c*e*g + b*d*i + a*f*h)
    return int(det)

# Calcular determinante de nuestra matriz
det = determinante_2x2(MATRIZ_CLAVE)
det_mod = det % 26

print(f"Matriz:")
print(MATRIZ_CLAVE)
print(f"\nDeterminante = {MATRIZ_CLAVE[0,0]}×{MATRIZ_CLAVE[1,1]} - {MATRIZ_CLAVE[0,1]}×{MATRIZ_CLAVE[1,0]}")
print(f"             = {MATRIZ_CLAVE[0,0]*MATRIZ_CLAVE[1,1]} - {MATRIZ_CLAVE[0,1]*MATRIZ_CLAVE[1,0]}")
print(f"             = {det}")
print(f"\nDeterminante mod 26 = {det_mod}")
print(f"MCD({det_mod}, 26) = {gcd(det_mod, 26)}")

if gcd(det_mod, 26) == 1:
    print(f"\n✓ La matriz es válida (determinante coprimo con 26)")
else:
    print(f"\n✗ La matriz NO es válida (determinante no coprimo con 26)")
````

### 3. Función: Matriz Inversa Modular (CORREGIDA)

````python
def matriz_cofactores_2x2(matriz):
    """
    Calcula la matriz de cofactores de una matriz 2×2.
    
    CORRECCIÓN APLICADA:
    Para matriz [[a, b], [c, d]], los cofactores son:
    [[d, -c],   ← CORRECTO
     [-b, a]]
    
    NO es [[d, -b], [-c, a]] ← Esto estaba mal antes
    
    Esto es parte del proceso para calcular la inversa.
    """
    return np.array([
        [matriz[1][1], -matriz[1][0]],   # [d, -c] ← CORREGIDO
        [-matriz[0][1], matriz[0][0]]    # [-b, a] ← CORREGIDO
    ])

def matriz_inversa_modular(matriz, modulo=26):
    """
    Calcula la matriz inversa modular.
    
    Proceso completo para matriz 2×2:
    1. Calcular determinante: det
    2. Verificar que MCD(det, 26) = 1
    3. Calcular inverso modular del determinante: det⁻¹
    4. Calcular matriz de cofactores (CORREGIDA)
    5. Transponer matriz de cofactores (= adjunta)
    6. Matriz inversa = det⁻¹ × adjunta (mod 26)
    
    Parámetros:
        matriz: matriz cuadrada (numpy array)
        modulo: módulo a usar (default 26)
    
    Retorna:
        Matriz inversa modular
    """
    n = len(matriz)
    
    # Paso 1: Calcular determinante
    if n == 2:
        det = determinante_2x2(matriz)
    elif n == 3:
        det = determinante_3x3(matriz)
    else:
        raise ValueError("Solo matrices 2×2 o 3×3 soportadas")
    
    # Determinante módulo 26
    det_mod = det % modulo
    
    # Paso 2: Verificar coprimalidad
    if gcd(det_mod, modulo) != 1:
        raise ValueError(f"Determinante {det_mod} no es coprimo con {modulo}")
    
    # Paso 3: Inverso modular del determinante
    det_inv = inverso_modular(det_mod, modulo)
    
    # Paso 4: Matriz de cofactores (CORREGIDA)
    if n == 2:
        cofactores = matriz_cofactores_2x2(matriz)
    else:
        raise NotImplementedError("Matriz 3×3 no implementada en esta celda")
    
    # Paso 5: Adjunta (transpuesta de cofactores)
    adjunta = cofactores.T
    
    # Paso 6: Matriz inversa
    inversa = (det_inv * adjunta) % modulo
    
    return inversa.astype(int)

# Calcular matriz inversa
matriz_inversa = matriz_inversa_modular(MATRIZ_CLAVE)

print("CÁLCULO DE LA MATRIZ INVERSA (CORREGIDO)")
print("="*60)
print(f"\nMatriz original K:")
print(MATRIZ_CLAVE)

print(f"\nPaso 1: det(K) = {det} ≡ {det_mod} (mod 26)")

det_inv = inverso_modular(det_mod, 26)
print(f"Paso 2: det⁻¹ = {det_mod}⁻¹ = {det_inv} (mod 26)")
print(f"        Verificación: {det_mod} × {det_inv} ≡ {(det_mod * det_inv) % 26} (mod 26) ✓")

cofactores = matriz_cofactores_2x2(MATRIZ_CLAVE)
print(f"\nPaso 3: Matriz de cofactores (CORREGIDA):")
print(cofactores)
print("        [[d, -c],  ← Formato correcto")
print("         [-b, a]]")

adjunta = cofactores.T
print(f"\nPaso 4: Adjunta (transpuesta):")
print(adjunta)

print(f"\nPaso 5: K⁻¹ = {det_inv} × adjunta (mod 26)")
print(f"\nMatriz inversa K⁻¹:")
print(matriz_inversa)

# Verificar que K × K⁻¹ = I (mod 26)
verificacion = (MATRIZ_CLAVE @ matriz_inversa) % 26
print(f"\n✓ Verificación: K × K⁻¹ (mod 26) =")
print(verificacion)
es_identidad = np.array_equal(verificacion, np.eye(2, dtype=int))
if es_identidad:
    print("\n✅ ¡ES LA MATRIZ IDENTIDAD! - Corrección exitosa")
else:
    print("\n❌ NO es la matriz identidad - Hay un error")
````

### 4. Funciones de Conversión

````python
def texto_a_numeros(texto):
    """
    Convierte texto a lista de números (A=0, B=1, ..., Z=25).
    
    Proceso:
    1. Limpiar texto (solo letras, mayúsculas)
    2. Para cada letra: ord(letra) - ord('A')
    
    Ejemplo:
        "HELP" → [7, 4, 11, 15]
        H=7, E=4, L=11, P=15
    """
    texto_limpio = ''.join(c.upper() for c in texto if c.isalpha())
    return [ord(c) - ord('A') for c in texto_limpio]

def numeros_a_texto(numeros):
    """
    Convierte lista de números a texto (0=A, 1=B, ..., 25=Z).
    
    Proceso:
    1. Para cada número: chr(numero + ord('A'))
    2. Unir todos los caracteres
    
    Ejemplo:
        [7, 4, 11, 15] → "HELP"
    """
    return ''.join(chr(n + ord('A')) for n in numeros)

# Convertir mensaje
numeros_mensaje = texto_a_numeros(MENSAJE)

print(f"Mensaje: {MENSAJE}")
print(f"Números: {numeros_mensaje}")
print(f"\nDetalle:")
for letra, num in zip(MENSAJE, numeros_mensaje):
    print(f"  {letra} → {num:2d}")
````

### 5. Función: Cifrar

````python
def cifrar_hill(texto, matriz_clave):
    """
    Cifra un texto usando el Cifrado de Hill.
    
    Proceso completo:
    1. Convertir texto a números
    2. Dividir en bloques del tamaño de la matriz
    3. Si es necesario, agregar padding (letra 'X' = 23)
    4. Para cada bloque:
       a. Crear vector columna con el bloque
       b. Multiplicar: C = K × M (mod 26)
       c. Guardar resultado
    5. Unir todos los bloques cifrados
    
    Fórmula matemática para matriz 2×2:
        [c1]   [k11 k12]   [m1]
        [c2] = [k21 k22] × [m2] (mod 26)
    
    Expandido:
        c1 = (k11×m1 + k12×m2) mod 26
        c2 = (k21×m1 + k22×m2) mod 26
    
    Parámetros:
        texto: mensaje a cifrar
        matriz_clave: matriz cuadrada (numpy array)
    
    Retorna:
        Tupla (texto_cifrado, lista_de_bloques_detallados)
    """
    n = len(matriz_clave)  # Tamaño de bloque
    
    # Convertir a números
    numeros = texto_a_numeros(texto)
    
    # Agregar padding si es necesario
    while len(numeros) % n != 0:
        numeros.append(23)  # 'X'
    
    # Cifrar por bloques
    texto_cifrado = []
    bloques_detallados = []
    
    for i in range(0, len(numeros), n):
        # Extraer bloque
        bloque = numeros[i:i+n]
        vector = np.array(bloque).reshape(n, 1)
        
        # Multiplicar matriz × vector (mod 26)
        resultado = (matriz_clave @ vector) % 26
        
        # Guardar
        bloque_cifrado = resultado.flatten().tolist()
        texto_cifrado.extend(bloque_cifrado)
        
        # Detalles
        bloques_detallados.append({
            'indice': i // n,
            'bloque_original': bloque,
            'texto_original': numeros_a_texto(bloque),
            'bloque_cifrado': bloque_cifrado,
            'texto_cifrado': numeros_a_texto(bloque_cifrado)
        })
    
    return numeros_a_texto(texto_cifrado), bloques_detallados

# CIFRAR EL MENSAJE
texto_cifrado, bloques_cifrado = cifrar_hill(MENSAJE, MATRIZ_CLAVE)

print("="*80)
print("PROCESO DE CIFRADO - HILL")
print("="*80)
print(f"\nMensaje original: {MENSAJE}")
print(f"En números: {numeros_mensaje}")
print(f"\nMatriz clave K:")
print(MATRIZ_CLAVE)

print(f"\nFórmula: C = K × M (mod 26)")
print("\nDonde:")
print("  K = matriz clave")
print("  M = vector del mensaje")
print("  C = vector cifrado")

# Mostrar detalles de cada bloque
for bloque in bloques_cifrado:
    print(f"\n" + "─"*80)
    print(f"Bloque {bloque['indice'] + 1}:")
    print("─"*80)
    print(f"Mensaje: {bloque['texto_original']} → {bloque['bloque_original']}")
    
    # Mostrar operación matricial
    m = bloque['bloque_original']
    c = bloque['bloque_cifrado']
    K = MATRIZ_CLAVE
    
    print(f"\nOperación:")
    for i in range(len(K)):
        print(f"│ {K[i][0]:2d} {K[i][1]:2d} │   │ {m[i]:2d} │", end="")
        if i == len(K) // 2:
            print(f"   =   │ {c[i]:2d} │", end="")
        else:
            print(f"       │ {c[i]:2d} │", end="")
        print(" (mod 26)")
    
    # Cálculos detallados
    print(f"\nCálculos:")
    for i in range(len(K)):
        operacion = " + ".join([f"{K[i][j]}×{m[j]}" for j in range(len(m))])
        suma = sum(K[i][j] * m[j] for j in range(len(m)))
        print(f"  c{i+1} = {operacion} = {suma} ≡ {c[i]} (mod 26)")
    
    print(f"\nCifrado: {bloque['texto_cifrado']} → {bloque['bloque_cifrado']}")

print("\n" + "="*80)
print(f"🔒 MENSAJE CIFRADO: {texto_cifrado}")
print(f"   Esperado: FW")
if texto_cifrado == "FW":
    print("   ✅ CORRECTO")
else:
    print(f"   ❌ ERROR")
print("="*80)
````

### 6. Función: Descifrar (AHORA FUNCIONA CORRECTAMENTE)

````python
def descifrar_hill(texto_cifrado, matriz_clave):
    """
    Descifra un texto usando el Cifrado de Hill.
    
    Proceso completo:
    1. Calcular la matriz inversa de la clave: K⁻¹ (AHORA CORREGIDA)
    2. Convertir texto cifrado a números
    3. Dividir en bloques
    4. Para cada bloque:
       a. Crear vector columna con el bloque cifrado
       b. Multiplicar: M = K⁻¹ × C (mod 26)
       c. Guardar resultado
    5. Unir todos los bloques descifrados
    
    Fórmula matemática:
        [m1]   [k⁻¹₁₁ k⁻¹₁₂]   [c1]
        [m2] = [k⁻¹₂₁ k⁻¹₂₂] × [c2] (mod 26)
    
    Parámetros:
        texto_cifrado: mensaje cifrado
        matriz_clave: matriz usada en el cifrado
    
    Retorna:
        Tupla (texto_descifrado, lista_de_bloques_detallados)
    """
    # Calcular matriz inversa (AHORA CORREGIDA)
    matriz_inv = matriz_inversa_modular(matriz_clave)
    n = len(matriz_clave)
    
    # Convertir a números
    numeros = texto_a_numeros(texto_cifrado)
    
    # Descifrar por bloques
    texto_descifrado = []
    bloques_detallados = []
    
    for i in range(0, len(numeros), n):
        # Extraer bloque
        bloque = numeros[i:i+n]
        vector = np.array(bloque).reshape(n, 1)
        
        # Multiplicar matriz_inversa × vector (mod 26)
        resultado = (matriz_inv @ vector) % 26
        
        # Guardar
        bloque_descifrado = resultado.flatten().tolist()
        texto_descifrado.extend(bloque_descifrado)
        
        # Detalles
        bloques_detallados.append({
            'indice': i // n,
            'bloque_cifrado': bloque,
            'texto_cifrado': numeros_a_texto(bloque),
            'bloque_descifrado': bloque_descifrado,
            'texto_descifrado': numeros_a_texto(bloque_descifrado)
        })
    
    return numeros_a_texto(texto_descifrado), bloques_detallados

# DESCIFRAR EL MENSAJE
texto_descifrado, bloques_descifrado = descifrar_hill(texto_cifrado, MATRIZ_CLAVE)

print("="*80)
print("PROCESO DE DESCIFRADO - HILL (CORREGIDO)")
print("="*80)
print(f"\nMensaje cifrado: {texto_cifrado}")
print(f"\nMatriz clave K:")
print(MATRIZ_CLAVE)
print(f"\nMatriz inversa K⁻¹ (CORREGIDA):")
print(matriz_inversa)

print(f"\nFórmula: M = K⁻¹ × C (mod 26)")

# Mostrar detalles de cada bloque
for bloque in bloques_descifrado:
    print(f"\n" + "─"*80)
    print(f"Bloque {bloque['indice'] + 1}:")
    print("─"*80)
    print(f"Cifrado: {bloque['texto_cifrado']} → {bloque['bloque_cifrado']}")
    
    # Mostrar operación matricial
    c = bloque['bloque_cifrado']
    m = bloque['bloque_descifrado']
    K_inv = matriz_inversa
    
    print(f"\nOperación:")
    for i in range(len(K_inv)):
        print(f"│ {K_inv[i][0]:2d} {K_inv[i][1]:2d} │   │ {c[i]:2d} │", end="")
        if i == len(K_inv) // 2:
            print(f"   =   │ {m[i]:2d} │", end="")
        else:
            print(f"       │ {m[i]:2d} │", end="")
        print(" (mod 26)")
    
    # Cálculos detallados
    print(f"\nCálculos:")
    for i in range(len(K_inv)):
        operacion = " + ".join([f"{K_inv[i][j]}×{c[j]}" for j in range(len(c))])
        suma = sum(K_inv[i][j] * c[j] for j in range(len(c)))
        print(f"  m{i+1} = {operacion} = {suma} ≡ {m[i]} (mod 26)")
    
    print(f"\nDescifrado: {bloque['texto_descifrado']} → {bloque['bloque_descifrado']}")

print("\n" + "="*80)
print(f"🔓 MENSAJE DESCIFRADO: {texto_descifrado}")
print(f"   Original: {MENSAJE}")
if texto_descifrado == MENSAJE:
    print("   ✅ CORRECTO - ¡El descifrado funciona!")
else:
    print(f"   ❌ ERROR - Se obtuvo {texto_descifrado}")
print("="*80)

# Verificación final
print(f"\n✅ Verificación: {MENSAJE} == {texto_descifrado} → {MENSAJE == texto_descifrado}")
````

### 📐 Explicación Matemática

Para matriz `[[a,b],[c,d]]`, los **cofactores** son:

- C₀₀ = d (menor de a)
- C₀₁ = -c (menor de b, con signo -)
- C₁₀ = -b (menor de c, con signo -)
- C₁₁ = a (menor de d)

Por lo tanto: `Cofactores = [[d,-c],[-b,a]]`
